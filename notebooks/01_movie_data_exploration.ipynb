## 5. Data Cleaning

```python
import pandas as pd

df = pd.read_csv("../data/IMDB-Movie-Dataset(2023-1951).csv")
df_clean = df.copy()

df_clean = df_clean.drop(columns=['Unnamed: 0'])

df_clean['year'] = df_clean['year'].where(
    df_clean['year'].str.match(r'^\d{4}$', na=False)
)

df_clean['year'] = pd.to_numeric(df_clean['year'], errors='coerce')

df_clean['genre'] = df_clean['genre'].replace(
    r'^\s*$', pd.NA, regex=True
)

df_clean['year'] = df_clean['year'].astype('Int64')

df_clean.info()

df_clean.isnull().sum()

df_clean.shape
```

```python
text_columns = ['movie_name', 'genre', 'overview', 'director', 'cast']

for col in text_columns:
    df_clean[col] = df_clean[col].str.strip()
```

```python
df_clean[df_clean['genre'].isna()][
    ['movie_name', 'overview', 'director', 'cast']
]
```

```python
df_clean['genre'] = df_clean['genre'].fillna('Unknown')
df_clean.isnull().sum()
```

```python
df_clean['tags'] = (
    df_clean['genre'].fillna('') + ' ' +
    df_clean['overview'].fillna('') + ' ' +
    df_clean['director'].fillna('') + ' ' +
    df_clean['cast'].fillna('')
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(
    stop_words='english'
)
```

```python
tfidf_matrix = tfidf.fit_transform(df_clean['tags'])
```

```python
tfidf_matrix.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(tfidf_matrix)
```

```python
similarity.shape
```

```python
movie_indices = pd.Series(
    df_clean.index,
    index=df_clean['movie_name']
).drop_duplicates()
```

```python
def recommend(movie_name, num_recommendations=10):
    if movie_name not in movie_indices:
        return f"Movie '{movie_name}' not found."

    movie_index = movie_indices[movie_name]

    similarity_scores = list(enumerate(similarity[movie_index]))

    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    similar_movies = similarity_scores[1:num_recommendations + 1]

    recommendations = []

    for index, score in similar_movies:
        recommendations.append(
            (df_clean.iloc[index]['movie_name'], round(score, 3))
        )

    return recommendations
```

```python
tfidf_weighted = TfidfVectorizer(
    stop_words='english'
)

tfidf_matrix_weighted = tfidf_weighted.fit_transform(
    df_clean['weighted_tags']
)

print("TF-IDF matrix shape:", tfidf_matrix_weighted.shape)
```

```python
similarity_weighted = cosine_similarity(
    tfidf_matrix_weighted
)

print("Similarity matrix shape:", similarity_weighted.shape)
```

```python
def recommend_weighted(movie_key, num_recommendations=10):
    if movie_key not in movie_indices:
        return f"Movie '{movie_key}' not found."

    movie_index = movie_indices[movie_key]

    similarity_scores = list(
        enumerate(similarity_weighted[movie_index])
    )

    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    similar_movies = similarity_scores[
        1:num_recommendations + 1
    ]

    recommendations = []

    for index, score in similar_movies:
        recommendations.append(
            (
                df_clean.iloc[index]['movie_key'],
                round(score, 3)
            )
        )

    return recommendations
```

```python
def recommend_weighted(movie_key, num_recommendations=10):
    if movie_key not in movie_indices:
        return f"Movie '{movie_key}' not found."

    movie_index = movie_indices[movie_key]

    similarity_scores = list(
        enumerate(similarity_weighted[movie_index])
    )

    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    recommendations = []

    for index, score in similarity_scores:
        if index == movie_index:
            continue

        if df_clean.iloc[index]['is_placeholder']:
            continue

        recommendations.append(
            (
                df_clean.iloc[index]['movie_key'],
                round(score, 3)
            )
        )

        if len(recommendations) == num_recommendations:
            break

    return recommendations
```
